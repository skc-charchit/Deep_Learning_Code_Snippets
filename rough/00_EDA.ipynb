# Histogram( Data Distribution)

#### Definition:
- A histogram is a graphical representation of the distribution of a continuous numerical dataset. It divides the range of data into intervals (bins) and shows the frequency of data points within each bin using bars.

#### Type of Data:

- Continuous numerical data (e.g., height, weight, test scores).
- Not suitable for categorical data.

### Bar Chart

#### Definition:
- A bar chart is a graphical representation of categorical data with rectangular bars, where the length (or height) of the bar represents the value or frequency of the category.

#### Purpose:
- To compare values across different categories.
- To visualize trends, proportions, or differences between groups.
- Useful for both single-series and stacked comparisons.

#### Type of Data:
- Categorical data (e.g., cities, products, departments).
- Can also be used with discrete numerical data.

# Scatter Plot

## Definition:
- A scatter plot is a graph that shows the relationship between two numerical variables by plotting individual data points on a Cartesian coordinate system. Each point represents a pair of values (x, y).

## Purpose:

- To visualize the relationship or correlation between two variables.

- To identify patterns such as positive/negative correlation, clusters, or outliers.

- Helpful in regression analysis or trend detection.

## Type of Data:

- Continuous numerical data for both x and y axes.

- Not suitable for categorical data (unless categories are encoded numerically).
