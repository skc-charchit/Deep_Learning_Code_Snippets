```python
inp = 5
h1 = 16
h2 = 32
o = 1

import torch
import torch.nn as nn
```

```python
# model building
class MyModel(nn.Module):
    def __init__(self,inp,out):
        super(MyModel,self).__init__()
        # self.inp = inp
        # self.o = o
        self.lin1 = nn.Linear(inp, 16)
        self.relu1 = nn.ReLU()
        self.lin2 = nn.Linear(16,32)
        self.relu2 = nn.ReLU()
        self.lin3 = nn.Linear(32,out)

    def forward(self, x):
        x = self.lin1(x)
        x = self.relu1(x)
        x = self.lin2(x)
        x = self.relu2(x)
        x = self.lin3(x)

        return x
    
```

```python
model = MyModel(5,1)
print(model)
```

```python
x = torch.tensor([1,2,3,4,5], dtype = torch.float32)
y_hat = model(x)
print(y_hat)
```
